# 4. EDA with SQL (SQLite)

```python
import sqlite3, pandas as pd
con=sqlite3.connect("my_data1.db")
pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/labs/module_2/data/Spacex.csv").to_sql("SPACEXTBL",con,if_exists="replace",index=False)
q=lambda s: pd.read_sql(s,con)
```

```python
q("SELECT DISTINCT Launch_Site FROM SPACEXTBL")
```

```python
q("SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5")
```

```python
q("SELECT SUM(PAYLOAD_MASS__KG_) FROM SPACEXTBL WHERE Customer='NASA (CRS)'")
```

```python
q("SELECT AVG(PAYLOAD_MASS__KG_) FROM SPACEXTBL WHERE Booster_Version='F9 v1.1'")
```

```python
q("SELECT MIN(Date) FROM SPACEXTBL WHERE Landing_Outcome='Success (ground pad)'")
```

```python
q("SELECT DISTINCT Booster_Version FROM SPACEXTBL WHERE Landing_Outcome='Success (drone ship)' AND PAYLOAD_MASS__KG_>4000 AND PAYLOAD_MASS__KG_<6000")
```

```python
q("SELECT Mission_Outcome, COUNT(*) FROM SPACEXTBL GROUP BY Mission_Outcome")
```

```python
q("SELECT DISTINCT Booster_Version FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_=(SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL)")
```

```python
q("SELECT substr(Date,6,2) Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE Landing_Outcome='Failure (drone ship)' AND substr(Date,0,5)='2015'")
```

```python
q("SELECT Landing_Outcome, COUNT(*) cnt FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY cnt DESC")
```
