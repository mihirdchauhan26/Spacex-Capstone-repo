# 2. Data Collection with Web Scraping
Scrape the Falcon 9 launch table from Wikipedia with BeautifulSoup.

```python
import requests, pandas as pd
from bs4 import BeautifulSoup
```

```python
url="https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"
soup=BeautifulSoup(requests.get(url).text,"html.parser")
soup.title.string
```

```python
tables=soup.find_all("table","wikitable plainrowheaders collapsible")
headers=[th.get_text(strip=True) for th in tables[0].find_all("th")]
headers[:12]
```

```python
rows=[]
for t in tables:
    for tr in t.find_all("tr")[1:]:
        cells=[c.get_text(strip=True) for c in tr.find_all(["th","td"])]
        if len(cells)>=8: rows.append(cells[:8])
df=pd.DataFrame(rows,columns=["FlightNo","DateTime","Version","LaunchSite","Payload","PayloadMass","Orbit","Customer"])
df.head()
```

```python
df.to_csv("../data/spacex_web_scraped.csv",index=False)
```
