# 5. EDA with Data Visualization

```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df=pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/api/dataset_part_2.csv")
```

```python
sns.catplot(x="FlightNumber",y="LaunchSite",hue="Class",data=df,aspect=4); plt.show()
```

```python
sns.catplot(x="PayloadMass",y="LaunchSite",hue="Class",data=df,aspect=4); plt.show()
```

```python
df.groupby("Orbit")["Class"].mean().plot(kind="bar"); plt.ylabel("Success rate"); plt.show()
```

```python
sns.catplot(x="FlightNumber",y="Orbit",hue="Class",data=df,aspect=4); plt.show()
```

```python
sns.catplot(x="PayloadMass",y="Orbit",hue="Class",data=df,aspect=4); plt.show()
```

```python
df["Year"]=pd.to_datetime(df["Date"]).dt.year
df.groupby("Year")["Class"].mean().plot(); plt.ylabel("Success rate"); plt.show()
```

```python
features=df[["FlightNumber","PayloadMass","Orbit","LaunchSite","Flights","GridFins","Reused","Legs","LandingPad","Block","ReusedCount","Serial"]]
features_one_hot=pd.get_dummies(features,columns=["Orbit","LaunchSite","LandingPad","Serial"]).astype(float)
features_one_hot.to_csv("../data/dataset_part_3.csv",index=False)
```
