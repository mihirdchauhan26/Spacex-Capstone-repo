# 3. Data Wrangling
Create the binary `Class` label: 1 = first stage landed, 0 = did not.

```python
import pandas as pd, numpy as np
df=pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/api/dataset_part_1.csv")
df.isnull().sum()/len(df)*100
```

```python
df["LaunchSite"].value_counts()
```

```python
df["Orbit"].value_counts()
```

```python
landing_outcomes=df["Outcome"].value_counts(); landing_outcomes
```

```python
bad_outcomes={o for o in landing_outcomes.keys() if o.startswith("False") or o=="None None"}
bad_outcomes
```

```python
df["Class"]=[0 if o in bad_outcomes else 1 for o in df["Outcome"]]
print(df["Class"].mean())
df.to_csv("../data/dataset_part_2.csv",index=False)
```
