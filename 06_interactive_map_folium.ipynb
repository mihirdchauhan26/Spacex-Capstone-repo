# 6. Interactive Map with Folium

```python
import folium, pandas as pd
from folium.plugins import MarkerCluster, MousePosition
from folium.features import DivIcon
df=pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0701EN-SkillsNetwork/labs/module_3/data/spacex_launch_geo.csv")
sites=df.groupby("Launch Site",as_index=False).first()[["Launch Site","Lat","Long"]]
sites
```

```python
m=folium.Map(location=[29.56,-95.09],zoom_start=5)
for _,r in sites.iterrows():
    loc=[r.Lat,r.Long]
    folium.Circle(loc,radius=1000,color="#d35400",fill=True).add_child(folium.Popup(r["Launch Site"])).add_to(m)
    folium.map.Marker(loc,icon=DivIcon(icon_size=(20,20),icon_anchor=(0,0),html=f"<b style='color:#d35400'>{r['Launch Site']}</b>")).add_to(m)
m
```

```python
df["color"]=df["class"].map({1:"green",0:"red"})
cl=MarkerCluster().add_to(m)
for _,r in df.iterrows():
    folium.Marker([r.Lat,r.Long],icon=folium.Icon(color="white",icon_color=r.color)).add_to(cl)
m
```

```python
MousePosition().add_to(m)
from math import radians,sin,cos,sqrt,atan2
def dist(a,b,c,d):
    R=6373.0; la1,lo1,la2,lo2=map(radians,[a,b,c,d]); x=sin((la2-la1)/2)**2+cos(la1)*cos(la2)*sin((lo2-lo1)/2)**2
    return R*2*atan2(sqrt(x),sqrt(1-x))
site=(28.563197,-80.576820)
coast=(28.56367,-80.57163)  # replace with your own picked point
d=dist(*site,*coast)
folium.PolyLine([site,coast],weight=1).add_to(m)
folium.Marker(coast,icon=DivIcon(html=f"<b>{d:.2f} KM</b>")).add_to(m)
m.save("../images/launch_site_map.html"); m
```
